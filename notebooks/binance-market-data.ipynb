## FOR SPOT MARKET

```python
def get_binance_candles(symbol, interval, start):
    base_url = 'https://api.binance.com/api/v3/klines'
    params = {
        'symbol': symbol,
        'interval': interval,
        'limit': 1000,
        'startTime': int(start.timestamp()*1000)
    }
    response = requests.get(base_url, params=params)
    candles = response.json()
    column_names = ['Open Time', 'Open', 'High', 'Low', 'Close', 'Volume', 'Close Time', 'Quote Asset Volume', 'Number of Trades', 'Taker Buy Base Asset Volume', 'Taker Buy Quote Asset Volume', 'Ignore']
    df = pd.DataFrame(candles, columns=column_names)
    df['Open Time'] = pd.to_datetime(df['Open Time'], unit='ms')
    df.insert(1, 'Symbol', symbol)
    return(df)
```

```python
def candels_download (symbol, interval, start,end):
    df = pd.DataFrame()
    date_end = dt.datetime.strptime(end, '%Y-%m-%d')
    start = dt.datetime.strptime(start, '%Y-%m-%d')
    for i in list(range(0,int(((date_end - start)*24).days/1000) + (1 if ((date_end - start)*24).days/1000 % 1 > 0 else 0))):
        candles_data = get_binance_candles(symbol, interval,start)
        df = pd.concat([df,candles_data])
        dates = df['Open Time'].tolist()
        dates_list = [dt.date() for dt in dates]
        start = dates[-1]
    return(df)
        
```

## FUNDING RATE

```python
def get_binance_Funding_rate(symbol, start):
    base_url = 'https://fapi.binance.com/fapi/v1/fundingRate'
    params = {
        'Symbol': symbol,
        'limit': 1000,
        'startTime': int(start.timestamp()*1000),
    }
    response = requests.get(base_url, params=params)
    data = response.json()
    column_names = ['name', 'fundingTime', 'fundingRate']
    df = pd.DataFrame(data, columns=column_names)
    df['fundingTime'] = pd.to_datetime(df['fundingTime'], unit='ms')
    df.insert(1, 'Symbol', symbol)
    df = df.iloc[:, 1:]
    return(df)
```

```python
def Funging_rate_download(symbol, start,end):
    df = pd.DataFrame()
    date_end = dt.datetime.strptime(end, '%Y-%m-%d')
    start = dt.datetime.strptime(start, '%Y-%m-%d')
    for i in list(range(0,int(((date_end - start)*24).days/1000) + (1 if ((date_end - start)*24).days/1000 % 1 > 0 else 0))):
        fund_rate = get_binance_Funding_rate(symbol,start)
        df = pd.concat([df,fund_rate])
        dates = df['fundingTime'].tolist()
        dates_list = [dt.date() for dt in dates]
        start = dates[-1]
    return(df)
```

```python
start = '2020-01-01'
end = '2024-02-28'
fund_rate = pd.DataFrame()

for symbol in currency_pairs:
    try:
        data = Funging_rate_download(symbol,start,end)
        fund_rate = pd.concat([fund_rate,data])
        print('OK_',symbol)
    except:
        print('NO_',symbol)
fund_rate.reset_index(drop=True).drop_duplicates().to_csv(r'data/funding_rates_from_2020-01-01_to_2024-02-28.csv')
```
